## Step 1: Install uv

- `uv` is a single, very fast tool that replaces `pip`, `virtualenv`, `pyenv`, and `poetry` for Python projects
- It manages the Python interpreter itself, project dependencies, and virtual environments all in one place
- The first step is to install `uv`



### Task: Install uv


- **Windows**: open PowerShell and run `powershell -ExecutionPolicy ByPass -c "irm https://astral.sh/uv/install.ps1 | iex"`
- **Mac/Linux**: Open terminal and run `curl -LsSf https://astral.sh/uv/install.sh | sh` (it is safe -- I promise ;))



### Task: Verify the install

Open a fresh terminal (PowerShell or Terminal) and run

```
uv --version
```

You should see something like this (the version number may be different for you):

```
$ uv --version
uv 0.9.7
```

Congratulations! You have successfully installed `uv` and are ready to move on to the next step!


## Step 2: Set up the course environment with `uv sync`

- With `uv` installed we are now ready to set up our Python environment
- `uv` keeps a dedicated virtual environment *per project*, described by a `pyproject.toml` file (and locked precisely in a `uv.lock` file)
- The course repository already ships with a `pyproject.toml` that lists every package we'll need this semester (numpy, pandas, matplotlib, networkx, and more)
- To install exactly those packages -- and only those packages -- we run `uv sync` from inside the repository


### Task: Run `uv sync`

Open a terminal, `cd` into the folder containing the course repository (the one with `pyproject.toml` in it), and run

```
uv sync
```

`uv` will download the right version of Python (if you don't already have it), create a `.venv` folder holding a virtual environment, and install every dependency listed in `pyproject.toml` at the exact version recorded in `uv.lock`



### Task: Launch Jupyter Lab with `uv run`

- `uv run <command>` executes `<command>` inside the project's virtual environment, without you needing to manually "activate" anything
- Launch Jupyter Lab by running

```
uv run jupyter lab
```

A web browser should pop open with the Jupyter Lab interface (should look similar to what I'm using)

### Task: run Python code in VSCode

- We can now run Python code in VSCode
- Open a new file in VSCode (File -> New File)
- Type the following code into the file:

```python

```

- Save the file as `hello.py` in a location you can find later
- Press `Ctrl + Shift + P` (or `Cmd + Shift + P` on Mac) to open the command palette
- Type "Python" and select "Python: Run Python File in Terminal"
- You should see the output of the code show up in a new terminal window within VS Code



### Task: run code with shift + Enter

Go back to `hello.py` and add the code

```python

```

Then, put your cursor on the line that contains `import numpy` and press shift and enter at the same time

This will start a new interactive Python session within your VS Code session and execute the code for you

It will also move your cursor down to the `n = 5` line. Press shift + enter again (repeatedly) to execute the remaining lines
