## What is Python?

- Python is a general purpose, high level programming language first released in 1991
- It emphasizes readable, expressive syntax -- Python code often reads almost like pseudocode
- On its own Python is not the fastest language, but its massive scientific ecosystem (`numpy`, `pandas`, `scipy`, and friends) lets us write code that is both clear to read and fast to run
- It is, by a wide margin, the most popular language for data science, scientific computing, and machine learning today

## Lists

- When doing numerical work, we often need to deal with multiple pieces of data at the same time
- In Python the default way of doing this is to use a `list`
- Lists are defined with `[` and `]` as below

````python
[1, 2, 3.14]  # a 3 element list
````

- Plain Python lists are general purpose containers, not built for linear algebra
- For matrices and vectorized math we reach for `numpy` arrays instead

````python
import numpy as np

np.array([1, 2, 3])  # a numpy array (vector)
````

````python
np.array([[1, 2, 3]])  # a 1x3 matrix
````

````python
np.array([[1, 2], [3, 4]])  # a 2x2 matrix
````

````python
np.array([
    [1, 2],
    [3, 4],
])  # another way to write a 2x2 matrix
````


### Accessing list/array items

- We can use `[N]` to access the Nth element
- We can also use `[i:j]` to access items `i` through `j` (exclusive of `j`)
- Finally we can use fancy indexing with a list of positions to access several elements at once


````python
[100, 101, 102, 103][1]
````

````python
[100, 101, 102, 103][1:4]
````

````python
np.array([100, 101, 102, 103])[[0, 2]]
````

- Note that Python starts counting at 0
- Also note that `-1` refers to the last element, `-2` to the second to last, and so on

````python
[1, 2, 3, 4][-1]
````

````python
[1, 2, 3, 4][-3]
````

## Dictionaries

- Very often in programming we want to be able to associate a key or name to a specific value
- One data type for doing that is a `dict`
- Dicts are created with the syntax `{key: value, ...}` where the `...` means we can repeat the pattern multiple times
- The keys and values can be of any (hashable, for keys) type


````python
{"x": 1, 2: "y", ("w", "z"): [1, 2, 3]}
````

````python
# use `[key]` to access the element with that key
{"x": 1, 2: "y", ("w", "z"): [1, 2, 3]}[2]
````

````python
{"x": 1, "y": 2}
````

- Dictionaries are often used for passing around groups of parameters
- A plain `dict` with string keys is a common way to bundle a handful of named values together
- We'll see examples later on

### Defining Functions

- Functions are used to execute a predefined set of operations
- Defining our own functions allows us to break programs into small, easily written and understood components
- We define functions using the syntax
```python
def name(arg1, arg2):
    # steps
    return result
```


````python
def mean(x):
    total = sum(x)
    n = len(x)
    return total / n
````

````python
mean([1, 2, 3])
````

````python
# mean of 1000 random samples from U[0,1] -- should be ~ 0.5
mean(np.random.rand(1000))
````


- If a function only contains one expression, you can also use a `lambda` (Python's anonymous-function shorthand):
```python
function_name = lambda arg1, arg2: # step
```


````python
add_two = lambda x: x + 2
````

````python
add_two(40)
````

## Comprehensions

- Python has a compact syntax -- the *comprehension* -- for building a list (or dict, or set) out of a loop and an expression
- A comprehension `[expr for item in iterable]` is equivalent to (but often faster than) building up a list with a `for` loop and `.append`


````python
[i ** 2 for i in range(1, 6)]  # squares of 1 through 5
````

````python
[i for i in range(20) if i % 3 == 0]  # comprehensions can filter with `if`
````

````python
{i: i ** 2 for i in range(5)}  # dict comprehension
````

````python
np.array([i ** 2 for i in range(1, 6)])  # wrap in np.array for vectorized follow-up math
````
