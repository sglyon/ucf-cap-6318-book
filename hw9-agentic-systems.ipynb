## Part C: Evaluation with Span Checking (30 points)

### 1. Deterministic Evaluators (10 points)

Create evaluators that check citations and validity.

```python
# TODO: Implement citation evaluators

class HasMinimumCitations(Evaluator):
    """Ensure answer has enough citations."""
    def __init__(self, min_citations: int = 2):
        self.min_citations = min_citations

    def evaluate(self, ctx: EvaluatorContext) -> EvaluationReason:
        answer: Answer = ctx.output
        has_enough = len(answer.citations) >= self.min_citations
        return EvaluationReason(
            value=has_enough,
            explanation=f"Has {len(answer.citations)} citations (need {self.min_citations})"
        )

class ValidCitations(Evaluator):
    """Check that all cited doc_ids exist."""
    def evaluate(self, ctx: EvaluatorContext) -> EvaluationReason:
        answer: Answer = ctx.output
        deps: KnowledgeBaseDeps = ctx.metadata['deps']

        invalid_citations = []
        for citation in answer.citations:
            if citation.doc_id not in deps.documents:
                invalid_citations.append(citation.doc_id)

        is_valid = len(invalid_citations) == 0
        return EvaluationReason(
            value=is_valid,
            explanation=f"Invalid citations: {invalid_citations}" if not is_valid else "All citations valid"
        )

print("✓ Deterministic evaluators defined")
```
