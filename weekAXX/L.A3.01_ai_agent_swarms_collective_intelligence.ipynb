## From Individual Intelligence to Collective Intelligence

### What We've Learned So Far

Over the past weeks, we've studied three distinct paradigms for computational agents:

**Rule-Based Agents (Weeks 6-7)**:
- Simple, deterministic rules: "move if unhappy"
- Local information only
- Fast and predictable
- Example: Schelling segregation model
  - Result: Stark segregation from mild preferences
  - Emergence: Collective pattern differs from individual intent

**Individual AI Agents (Weeks A1-A2)**:
- Flexible reasoning via LLMs
- Can use tools and access knowledge
- Adaptive and context-aware
- Example: Network analysis agent
  - Can answer complex questions
  - Chains multiple tool calls

**Networks (Weeks 3-5)**:
- Structure matters for information flow
- Small-world networks enable fast spreading
- Hub nodes have outsized influence
- Path structure determines reachability

### The Next Frontier: Swarms of AI Agents

Now we ask: **What happens when we combine these ideas?**

What if we had:
- **Many** AI agents (like ABMs have many agents)
- Each with **intelligence** (reasoning, tool use, adaptation)
- Connected via **network structures** (communication topologies)
- Working toward **collective goals** (distributed problem-solving)

This is **swarm intelligence** with AI agents - and it creates fundamentally new capabilities.

### Why Swarms?

Consider these scenarios:

**Scientific Discovery**:
- Single AI agent reads a paper → summarizes it
- Swarm of 50 AI agents:
  - 10 agents each read different papers in a field
  - 5 agents identify contradictions across papers
  - 15 agents propose hypotheses to resolve contradictions
  - 10 agents design experiments to test hypotheses
  - 5 agents synthesize findings into new theory
  - 5 agents critique the theory from different perspectives

**Software Engineering**:
- Single AI agent writes code → may have bugs
- Swarm approach:
  - Architect agents design system structure
  - Developer agents implement components
  - Tester agents write and run tests
  - Reviewer agents identify issues
  - Optimizer agents improve performance
  - All communicate via shared codebase and issue tracker

**Market Analysis**:
- Single agent analyzes market → one perspective
- Swarm:
  - Agents with different economic theories (Keynesian, Austrian, Behavioral)
  - Each analyzes same data from their framework
  - Debate and synthesis reveals robust insights
  - Diversity of perspective reduces blind spots

**The Core Insight**: 

Just as simple agents in ABMs create complex emergent behavior, **intelligent agents can create emergent intelligence** beyond what any single agent can achieve.

## Comparing Agent Paradigms: A Deeper Look

### Schelling Model Revisited

Let's recall the Schelling segregation model:

**Setup**:
- 25×25 grid
- 500 agents (250 orange, 250 blue)
- Each agent has a "happiness threshold" (e.g., N=3)
- Rule: "If fewer than N of my 8 neighbors are my type, move to random empty location"

**Result**:
- Stark segregation emerges
- Large homogeneous regions form
- This happens even with mild preference (N=3 means 37.5% same-type is acceptable!)

**Key Features**:
- Local information only (8 neighbors)
- No global coordination
- Deterministic rule application
- Fast execution (millions of steps per second)
- Emergence: macro segregation from micro preferences

### What if Agents Could Reason?

Imagine replacing rule-based agents with AI agents:

**AI Agent Schelling**:
- Each agent can:
  - Consider not just current neighbors but future trends
  - Reason about reputation effects of moving
  - Communicate with potential neighbors before moving
  - Form coalitions to collectively improve neighborhood
  - Learn from past moves and adapt strategy
  - Consider multiple objectives (diversity, affordability, schools)

**Potential Outcomes**:
- More complex spatial patterns (not just segregation)
- Emergent negotiation protocols
- Formation of diverse neighborhoods through active coordination
- Or possibly even more segregation if agents reason strategically

We don't know a priori - and that's what makes it interesting!

### The Computational Trade-off

| Aspect | Rule-Based Swarm | AI Agent Swarm |
|--------|------------------|----------------|
| **Agent Complexity** | Simple rules | Rich reasoning |
| **Speed** | ~1M steps/sec | ~10 steps/sec |
| **Swarm Size** | 10,000+ agents | 10-100 agents |
| **Behavior** | Programmed | Emergent from LLM |
| **Adaptation** | Fixed rules | Learn and adjust |
| **Communication** | Implicit (shared space) | Explicit (messages) |
| **Coordination** | Local only | Can be global |
| **Emergence** | Spatial/statistical | Cognitive/strategic |
| **Predictability** | High (deterministic) | Lower (stochastic LLM) |
| **Cost** | Minimal | $0.01-$1 per agent-step |

**The Sweet Spot**: 
- Use rule-based ABMs for large populations and statistical patterns
- Use AI agent swarms for problems requiring reasoning, strategy, and coordination
- Use hybrid approaches for realistic simulations

### Core Agent Infrastructure

We need infrastructure for:
1. Creating AI agents with identities
2. Enabling agent-to-agent communication

```julia
@dataclass
class SwarmAgent:
    """
    An AI agent in a swarm with identity, memory, and communication capabilities.
    
    Fields:
        id: Unique identifier
        role: Agent's role/specialty in the swarm
        system_prompt: Instructions defining agent's behavior
        memory: Conversation history (messages sent/received)
        state: Current internal state (beliefs, knowledge, etc.)
        agent: PydanticAI agent instance for LLM communication
    """
    id: str
    role: str
    system_prompt: str
    memory: List[Dict[str, str]] = field(default_factory=list)
    state: Dict[str, Any] = field(default_factory=dict)
    agent: Optional[Agent] = None
    
    def __post_init__(self):
        """Initialize the PydanticAI agent after dataclass creation."""
        if self.agent is None:
            self.agent = Agent(
                'anthropic:claude-3-5-sonnet-20241022',
                result_type=str,
                system_prompt=self.system_prompt
            )
    
    def __repr__(self):
        return f"SwarmAgent({self.id}, {self.role})"
```

```julia
"""
    call_claude(agent, message; model="claude-3-5-sonnet-20241022")

Make an API call for an agent, maintaining their conversation history.
"""
function call_claude(agent::SwarmAgent, message::String; model="claude-3-5-sonnet-20241022")
    url = "https://api.anthropic.com/v1/messages"
    
    headers = [
        "x-api-key" => ANTHROPIC_API_KEY,
        "anthropic-version" => "2023-06-01",
        "content-type" => "application/json"
    ]
    
    # Add user message to memory
    push!(agent.memory, Dict("role" => "user", "content" => message))
    
    body = JSON3.write(Dict(
        "model" => model,
        "max_tokens" => 1024,
        "system" => agent.system_prompt,
        "messages" => agent.memory
    ))
    
    response = HTTP.post(url, headers, body)
    result = JSON3.read(String(response.body))
    
    response_text = result.content[1].text
    
    # Add assistant response to memory
    push!(agent.memory, Dict("role" => "assistant", "content" => response_text))
    
    return response_text
end
```

```julia
"""
    Swarm

A collection of agents with a communication network.

# Fields
- `agents`: Dict mapping agent IDs to SwarmAgent objects
- `network`: Graph structure defining who can communicate with whom
- `message_log`: History of all inter-agent communications
"""
mutable struct Swarm
    agents::Dict{String, SwarmAgent}
    network::SimpleGraph
    message_log::Vector{Dict}
end

function Swarm(agents::Vector{SwarmAgent}, network::SimpleGraph)
    agent_dict = Dict(agent.id => agent for agent in agents)
    Swarm(agent_dict, network, Dict[])
end

"""
    send_message(swarm, from_id, to_id, message)

Send a message from one agent to another (if connected in network).
"""
function send_message!(swarm::Swarm, from_id::String, to_id::String, message::String)
    # Get agent indices
    agent_ids = sort(collect(keys(swarm.agents)))
    from_idx = findfirst(==(from_id), agent_ids)
    to_idx = findfirst(==(to_id), agent_ids)
    
    # Check if communication is allowed by network
    if !has_edge(swarm.network, from_idx, to_idx)
        error("Agents $from_id and $to_id are not connected in the network")
    end
    
    # Log the message
    push!(swarm.message_log, Dict(
        "from" => from_id,
        "to" => to_id,
        "message" => message,
        "timestamp" => length(swarm.message_log) + 1
    ))
    
    # Add to recipient's memory
    formatted_message = "Message from $from_id: $message"
    push!(swarm.agents[to_id].memory, Dict(
        "role" => "user",
        "content" => formatted_message
    ))
    
    return formatted_message
end
```

## Communication Topologies: Networks for Swarms

### Why Network Structure Matters

The communication topology determines:
- **Information flow**: How quickly ideas spread
- **Coordination**: Ability to achieve consensus
- **Specialization**: Formation of sub-groups
- **Robustness**: Resilience to agent failures

Let's explore different topologies:

### 1. Fully Connected (Complete Graph)

Every agent can communicate with every other agent.

**Pros**:
- Maximum information sharing
- Fast consensus
- No bottlenecks

**Cons**:
- O(N²) communication channels
- Expensive at scale
- Information overload
- No emergent structure

**Best for**: Small swarms (5-10 agents), brainstorming, rapid coordination

```julia
# Create a fully connected network
n_agents = 6
fully_connected = complete_graph(n_agents)

println("Fully connected network:")
println("  Nodes: ", nv(fully_connected))
println("  Edges: ", ne(fully_connected))
println("  Density: ", round(density(fully_connected), digits=3))

# Visualize
using GraphPlot
gplot(fully_connected, nodelabel=1:n_agents)
```

### 2. Ring (Cycle Graph)

Agents arranged in a circle, each connected to two neighbors.

**Pros**:
- Efficient communication (2N edges)
- Information eventually reaches everyone
- Natural for sequential processing

**Cons**:
- Slow information propagation (O(N/2) steps to reach farthest agent)
- Fragile (breaking one edge disconnects network)
- No hubs

**Best for**: Pipeline workflows, sequential review processes

```julia
ring = cycle_graph(n_agents)

println("Ring network:")
println("  Nodes: ", nv(ring))
println("  Edges: ", ne(ring))
println("  Average distance: ", round(mean(gdistances(ring, 1)), digits=2))

gplot(ring, nodelabel=1:n_agents)
```

### 3. Star (Hub-and-Spoke)

One central hub connected to all others; periphery agents not connected to each other.

**Pros**:
- Efficient for centralized coordination
- Hub can aggregate and synthesize
- Low communication overhead (2N-2 edges)

**Cons**:
- Hub is bottleneck and single point of failure
- No peripheral collaboration
- Hub can become overwhelmed

**Best for**: Hierarchical tasks, manager-worker patterns, centralized control

```julia
star = star_graph(n_agents)

println("Star network:")
println("  Nodes: ", nv(star))
println("  Edges: ", ne(star))
println("  Hub degree: ", degree(star, 1))
println("  Peripheral degree: ", degree(star, 2))

gplot(star, nodelabel=1:n_agents)
```

### 4. Small-World Network

Mostly local connections with a few long-range "shortcuts".

**Pros**:
- High clustering (local collaboration)
- Short path lengths (global connectivity)
- Balances local and global communication
- Robust to failures

**Cons**:
- More complex to design
- Shortcuts may not be obvious

**Best for**: Large swarms, realistic social networks, balanced exploration/exploitation

Small-world networks are arguably the most interesting for AI swarms because they mirror real social and organizational structures.

```julia
# Watts-Strogatz model: start with ring, rewire edges with probability p
using Graphs: watts_strogatz

small_world = watts_strogatz(n_agents, 2, 0.3)  # k=2 neighbors, p=0.3 rewiring prob

println("Small-world network:")
println("  Nodes: ", nv(small_world))
println("  Edges: ", ne(small_world))
println("  Clustering: ", round(global_clustering_coefficient(small_world), digits=3))
println("  Average distance: ", round(mean(gdistances(small_world, 1)), digits=2))

gplot(small_world, nodelabel=1:n_agents)
```

### 5. Scale-Free Network (Preferential Attachment)

A few highly connected hubs, many low-degree nodes. Power law degree distribution.

**Pros**:
- Reflects real social networks (influencers and followers)
- Efficient information broadcasting via hubs
- Robust to random failures

**Cons**:
- Vulnerable to targeted attacks on hubs
- Hubs can become bottlenecks
- Unequal communication load

**Best for**: Mimicking social influence, viral spreading, hierarchical expertise

```julia
# Barabási-Albert model: preferential attachment
using Graphs: barabasi_albert

scale_free = barabasi_albert(n_agents, 2)  # Add each node with 2 edges

println("Scale-free network:")
println("  Nodes: ", nv(scale_free))
println("  Edges: ", ne(scale_free))
degrees = degree(scale_free)
println("  Max degree: ", maximum(degrees))
println("  Min degree: ", minimum(degrees))

gplot(scale_free, nodelabel=1:n_agents)
```

### Comparing Topologies

Let's create a comparison for a larger network:

```julia
# Compare topologies for N=20 agents
N = 20

networks = Dict(
    "Complete" => complete_graph(N),
    "Ring" => cycle_graph(N),
    "Star" => star_graph(N),
    "Small-World" => watts_strogatz(N, 4, 0.3),
    "Scale-Free" => barabasi_albert(N, 2)
)

comparison = DataFrame(
    Topology = String[],
    Edges = Int[],
    Density = Float64[],
    AvgDegree = Float64[],
    AvgDistance = Float64[],
    Clustering = Float64[]
)

for (name, net) in networks
    push!(comparison, (
        name,
        ne(net),
        round(density(net), digits=3),
        round(mean(degree(net)), digits=2),
        round(mean([mean(gdistances(net, i)) for i in 1:nv(net)]), digits=2),
        round(global_clustering_coefficient(net), digits=3)
    ))
end

sort!(comparison, :AvgDistance)
comparison
```

### Key Insights

From this comparison:

**Complete graph**:
- Maximum edges (190 = N(N-1)/2)
- Shortest average distance (1.0 - everyone is neighbors)
- Low clustering (trivially high, but no interesting structure)

**Ring**:
- Minimal edges (20 = N)
- Longest average distance (~5 steps)
- Zero clustering (no triangles)

**Star**:
- Very efficient (19 = N-1 edges)
- Short distance (1.9 average - everything goes through hub)
- No clustering among periphery

**Small-world**:
- Moderate edges (40 = 2N)
- Good distance (shorter than ring)
- High clustering (local groups)
- **Best balance** for most swarm applications

**Scale-free**:
- Similar edges to small-world
- Shorter distances (hubs are shortcuts)
- Lower clustering
- Realistic for social dynamics

**Design Principle**: Choose topology based on your task:
- Need fast consensus? → Complete or Star
- Want emergent specialization? → Small-world or Scale-free
- Sequential workflow? → Ring or Path
- Realistic social simulation? → Scale-free

## Distributed Problem Solving

### Parallel Hypothesis Generation

One of the most powerful applications of AI swarms is **parallel exploration** of solution spaces.

**The Pattern**:
1. Pose a problem to the swarm
2. Each agent generates hypotheses/solutions independently
3. Agents share and critique each other's ideas
4. Swarm converges on best solution(s)

This mirrors human brainstorming but at scale.

Let's build a simple example: **solving a network optimization problem**.

### Example: Network Vulnerability Analysis

**Problem**: Given a network, which node should we protect to maximize network resilience?

This is a complex problem because:
- Multiple metrics matter (degree centrality, betweenness, clustering)
- Trade-offs exist (high degree ≠ highest betweenness)
- Context matters (what kind of failure? what kind of flow?)

Let's see how a swarm approaches this:

```julia
# Create a test network
Random.seed!(42)
problem_network = watts_strogatz(15, 3, 0.2)

println("Problem network:")
println("  Nodes: ", nv(problem_network))
println("  Edges: ", ne(problem_network))

# Visualize with node labels
gplot(problem_network, nodelabel=1:15)
```

```julia
# Create agents with different analytical perspectives

analyst1 = SwarmAgent(
    "analyst_1",
    "Degree Centrality Specialist",
    """
    You are a network analyst specializing in degree centrality.
    You believe that nodes with the most connections are most critical.
    When analyzing networks, you focus on identifying high-degree hubs.
    Be concise but explain your reasoning.
    """
)

analyst2 = SwarmAgent(
    "analyst_2",
    "Betweenness Centrality Specialist",
    """
    You are a network analyst specializing in betweenness centrality.
    You believe that nodes that lie on many shortest paths are most critical.
    These 'bridge' nodes control information flow.
    Be concise but explain your reasoning.
    """
)

analyst3 = SwarmAgent(
    "analyst_3",
    "Clustering Specialist",
    """
    You are a network analyst specializing in local clustering.
    You believe that nodes in highly clustered regions are critical because
    they maintain community structure.
    Be concise but explain your reasoning.
    """
)

synthesizer = SwarmAgent(
    "synthesizer",
    "Synthesis Expert",
    """
    You are a senior analyst who synthesizes different perspectives.
    Review the analyses from other agents and provide a balanced recommendation.
    Consider trade-offs and context.
    Be concise and decisive.
    """
)

agents = [analyst1, analyst2, analyst3, synthesizer]

# Create star network: analysts report to synthesizer
comm_network = star_graph(4)  # synthesizer at center (node 1)

swarm = Swarm(agents, comm_network)

println("Created swarm with $(length(agents)) agents")
println("Communication network: star topology")
```

## Emergent Specialization and Division of Labor

### From Assignment to Emergence

In the previous example, we **assigned** roles to agents explicitly:
- Agent 1: degree specialist
- Agent 2: betweenness specialist
- etc.

But in natural swarms (ants, bees, humans), specialization often **emerges** without central assignment:
- Workers gravitate toward tasks they're good at
- Experience reinforces specialization
- Market-like dynamics allocate labor efficiently

Can AI agent swarms develop emergent specialization?

### The Skill Development Model

Let's build a model where:
1. Agents start as generalists (same initial capabilities)
2. They perform tasks and accumulate experience
3. Success reinforces skill in that domain
4. Agents preferentially take tasks matching their skills
5. Specialization emerges from this feedback loop

This mirrors how scientific research teams self-organize:

```julia
# Define task types for research
TASK_TYPES = [
    "literature_review",
    "hypothesis_generation",
    "experiment_design",
    "data_analysis",
    "writing",
    "peer_review"
]

# Create agents with initially equal (low) skill in all domains
function create_research_agent(id::String)
    agent = SwarmAgent(
        id,
        "Researcher",
        """
        You are a research scientist. You will be assigned tasks of different types.
        Your performance improves with experience in each task type.
        You prefer tasks you're good at but are willing to try new things.
        """
    )
    
    # Initialize skill levels (0.0 to 1.0)
    agent.state["skills"] = Dict(task => 0.3 + 0.1*randn() for task in TASK_TYPES)
    agent.state["experience"] = Dict(task => 0 for task in TASK_TYPES)
    agent.state["tasks_completed"] = 0
    
    return agent
end

println("Task types defined: ", join(TASK_TYPES, ", "))
```

```julia
"""
    assign_task(swarm, task_type)

Assign a task to the agent with highest skill in that domain.
Simulate task performance and skill improvement.
"""
function assign_task!(swarm::Swarm, task_type::String)
    # Find agent with highest skill in this task
    agent_skills = [(id, agent.state["skills"][task_type]) 
                    for (id, agent) in swarm.agents]
    best_agent_id = first(sort(agent_skills, by=x->x[2], rev=true))[1]
    
    agent = swarm.agents[best_agent_id]
    
    # Simulate task performance (skill + randomness)
    current_skill = agent.state["skills"][task_type]
    performance = current_skill + 0.2 * randn()
    performance = clamp(performance, 0.0, 1.0)
    
    # Update experience and skill
    agent.state["experience"][task_type] += 1
    agent.state["tasks_completed"] += 1
    
    # Skill improves with experience (learning curve)
    learning_rate = 0.1
    agent.state["skills"][task_type] = min(
        1.0,
        agent.state["skills"][task_type] + learning_rate * (1 - current_skill)
    )
    
    return Dict(
        "agent" => best_agent_id,
        "task" => task_type,
        "performance" => performance,
        "skill_before" => current_skill,
        "skill_after" => agent.state["skills"][task_type]
    )
end
```

```julia
# Create a swarm of 6 researchers
Random.seed!(42)
n_researchers = 6

researchers = [create_research_agent("researcher_$i") for i in 1:n_researchers]

# Create a small-world network for collaboration
collab_network = watts_strogatz(n_researchers, 2, 0.3)

research_swarm = Swarm(researchers, collab_network)

# Print initial skills
println("Initial skill distribution (all agents start similar):")
for (id, agent) in research_swarm.agents
    avg_skill = mean(values(agent.state["skills"]))
    println("  $id: average skill = $(round(avg_skill, digits=3))")
end
```

```julia
# Simulate 100 tasks
n_tasks = 100
task_log = []

println("\nSimulating $n_tasks tasks...")

for i in 1:n_tasks
    # Random task type
    task_type = rand(TASK_TYPES)
    
    # Assign to best available agent
    result = assign_task!(research_swarm, task_type)
    push!(task_log, result)
    
    if i % 20 == 0
        println("  Completed $i tasks...")
    end
end

println("Simulation complete!")
```

```julia
# Analyze emergent specialization
println("\nFinal agent specializations:")
println("=" ^ 80)

for (id, agent) in sort(collect(research_swarm.agents), by=x->x[1])
    println("\n$id:")
    
    # Find specialization (highest skill)
    skills = agent.state["skills"]
    sorted_skills = sort(collect(skills), by=x->x[2], rev=true)
    
    println("  Primary specialty: $(sorted_skills[1][1]) (skill: $(round(sorted_skills[1][2], digits=3)))")
    println("  Secondary: $(sorted_skills[2][1]) (skill: $(round(sorted_skills[2][2], digits=3)))")
    println("  Tasks completed: $(agent.state["tasks_completed"])")
    
    # Show experience distribution
    exp = agent.state["experience"]
    sorted_exp = sort(collect(exp), by=x->x[2], rev=true)
    println("  Experience profile: $(join(["$(e[1]): $(e[2])" for e in sorted_exp[1:3]], ", "))")
end
```

### Insights from Emergent Specialization

What we've observed:

**1. Differentiation from Uniformity**:
- All agents started with similar skills
- After 100 tasks, clear specializations emerged
- No central planner assigned roles
- Market-like dynamics (best agent gets task) drove differentiation

**2. Positive Feedback Loop**:
- Agent good at task X gets assigned more X tasks
- More X tasks → more experience → better at X
- Better at X → even more likely to get X tasks
- This is the "rich get richer" dynamic we saw in preferential attachment

**3. Efficiency vs. Exploration**:
- Our allocation favored efficiency (best agent for each task)
- But this limits exploration (agents rarely try new tasks)
- Real systems balance exploitation and exploration

**4. Path Dependence**:
- Early random variations mattered
- Agent who happened to do well at first writing task became the "writer"
- Different random seed → different specializations

**Connection to Economics**:
- Adam Smith's division of labor: specialization increases productivity
- Comparative advantage: agents specialize where they have relative edge
- Self-organizing markets allocate resources efficiently

**Connection to ABMs**:
- Similar to money model: inequality emerges from equal starts
- But here inequality is functional (specialization is good!)
- Emergence at task allocation level, not spatial level

### A Realistic Application

Let's bring everything together in a realistic use case: **a swarm of AI agents collaborating on scientific research**.

**Scenario**: We want to understand the relationship between network structure and innovation diffusion.

**Swarm Composition**:
1. **Literature Agents** (2): Survey papers, extract key findings
2. **Hypothesis Agents** (3): Propose theories and mechanisms
3. **Simulation Agent** (1): Run computational experiments
4. **Analysis Agent** (1): Analyze simulation results
5. **Critique Agent** (1): Identify limitations and alternative explanations
6. **Synthesis Agent** (1): Integrate findings into coherent narrative

**Network Structure**: Small-world (mimics academic collaboration)

**Workflow**:
1. Literature agents summarize relevant research
2. Hypothesis agents propose mechanisms
3. Simulation agent tests hypotheses
4. Analysis agent interprets results
5. Critique agent challenges conclusions
6. Synthesis agent produces final report

This is a simplified version of real research collaboration!

```julia
# Create scientific discovery swarm

lit_agent_1 = SwarmAgent(
    "lit_1",
    "Literature Reviewer",
    """
    You are a literature review specialist in network science and innovation.
    Summarize key findings about how network structure affects innovation spread.
    Focus on empirical results and established theories.
    Be concise and cite key insights.
    """
)

hyp_agent_1 = SwarmAgent(
    "hyp_1",
    "Hypothesis Generator",
    """
    You generate hypotheses about social and economic phenomena.
    Based on literature and intuition, propose testable mechanisms.
    Be creative but grounded in theory.
    State hypotheses clearly and concisely.
    """
)

hyp_agent_2 = SwarmAgent(
    "hyp_2",
    "Alternative Hypothesis Generator",
    """
    You specialize in alternative explanations and competing theories.
    When others propose mechanisms, you suggest alternatives.
    Consider institutional, behavioral, and structural factors.
    Be constructively critical.
    """
)

sim_agent = SwarmAgent(
    "sim_1",
    "Simulation Designer",
    """
    You design computational experiments to test hypotheses.
    Specify simulation parameters and expected outcomes.
    Think about what evidence would support or refute hypotheses.
    Be precise about measurement.
    """
)

analysis_agent = SwarmAgent(
    "analysis_1",
    "Data Analyst",
    """
    You analyze simulation results and draw conclusions.
    Look for patterns, test statistical significance, interpret findings.
    Connect results back to hypotheses.
    Be rigorous and cautious about claims.
    """
)

critique_agent = SwarmAgent(
    "critique_1",
    "Research Critic",
    """
    You identify limitations, confounds, and alternative explanations.
    Challenge conclusions and suggest additional tests.
    Consider what might be missing or overlooked.
    Be skeptical but constructive.
    """
)

synthesis_agent_final = SwarmAgent(
    "synthesis_1",
    "Senior Researcher",
    """
    You synthesize findings into a coherent research narrative.
    Integrate literature, hypotheses, results, and critiques.
    Write a mini research summary with:
    - Research question
    - Key hypothesis
    - Main finding
    - Limitations
    - Future directions
    Keep it under 200 words.
    """
)

discovery_agents = [
    lit_agent_1,
    hyp_agent_1,
    hyp_agent_2,
    sim_agent,
    analysis_agent,
    critique_agent,
    synthesis_agent_final
]

# Create small-world collaboration network
collab_net = watts_strogatz(length(discovery_agents), 3, 0.3)

discovery_swarm = Swarm(discovery_agents, collab_net)

println("Created scientific discovery swarm with $(length(discovery_agents)) agents")
println("Collaboration network: small-world topology")
```

### Running the Research Process

Now we orchestrate the research workflow. Each step builds on previous outputs:

**Note**: This would make multiple API calls and cost money. For the lecture, we'll show the structure without running all steps.

```julia
# Research workflow (structure only - not executed to save API costs)

function run_discovery_workflow(swarm::Swarm)
    """
    Execute full research workflow through the swarm.
    """
    
    println("PHASE 1: Literature Review")
    println("=" ^ 80)
    
    lit_prompt = """
    Summarize key findings about how network structure (clustering, path length,
    degree distribution) affects innovation diffusion.
    
    What do we know from prior research? Be concise (under 150 words).
    """
    
    lit_summary = call_claude(swarm.agents["lit_1"], lit_prompt)
    println(lit_summary)
    println()
    
    # -----------------------------------------------------------------------
    
    println("PHASE 2: Hypothesis Generation")
    println("=" ^ 80)
    
    hyp_prompt = """
    Based on this literature review:
    
    $(lit_summary)
    
    Propose a specific, testable hypothesis about how one network structural
    feature affects innovation spread speed. Be precise.
    """
    
    hypothesis = call_claude(swarm.agents["hyp_1"], hyp_prompt)
    println("Primary hypothesis:")
    println(hypothesis)
    println()
    
    # Get alternative hypothesis
    alt_prompt = """
    A colleague proposed:
    
    $(hypothesis)
    
    What alternative mechanism or competing hypothesis should we also consider?
    """
    
    alt_hypothesis = call_claude(swarm.agents["hyp_2"], alt_prompt)
    println("Alternative hypothesis:")
    println(alt_hypothesis)
    println()
    
    # -----------------------------------------------------------------------
    
    println("PHASE 3: Simulation Design")
    println("=" ^ 80)
    
    sim_prompt = """
    We want to test these hypotheses:
    
    PRIMARY: $(hypothesis)
    
    ALTERNATIVE: $(alt_hypothesis)
    
    Design a simulation to test them. Specify:
    1. Network types to compare
    2. Innovation diffusion mechanism
    3. Key metrics to measure
    4. What results would support each hypothesis
    """
    
    sim_design = call_claude(swarm.agents["sim_1"], sim_prompt)
    println(sim_design)
    println()
    
    # -----------------------------------------------------------------------
    
    println("PHASE 4: Simulated Results (mock data)")
    println("=" ^ 80)
    
    # In reality, we'd run actual simulations here
    # For demonstration, we'll provide mock results
    mock_results = """
    Simulation results (100 runs each):
    
    - Small-world network: Innovation reached 80% of nodes in avg 12.3 steps (SD: 2.1)
    - Random network: Innovation reached 80% in avg 8.7 steps (SD: 1.8)
    - Scale-free network: Innovation reached 80% in avg 6.2 steps (SD: 2.4)
    
    Statistical tests: Scale-free significantly faster than small-world (p < 0.001)
    Random intermediate and significantly different from both (p < 0.01)
    """
    
    println(mock_results)
    println()
    
    # -----------------------------------------------------------------------
    
    println("PHASE 5: Analysis")
    println("=" ^ 80)
    
    analysis_prompt = """
    Here are simulation results:
    
    $(mock_results)
    
    Recall the hypotheses being tested:
    PRIMARY: $(hypothesis)
    ALTERNATIVE: $(alt_hypothesis)
    
    Interpret these results. Which hypothesis is supported? Why?
    """
    
    analysis = call_claude(swarm.agents["analysis_1"], analysis_prompt)
    println(analysis)
    println()
    
    # -----------------------------------------------------------------------
    
    println("PHASE 6: Critique")
    println("=" ^ 80)
    
    critique_prompt = """
    Analysis:
    $(analysis)
    
    Simulation design:
    $(sim_design)
    
    What are the key limitations? What alternative explanations exist?
    What would strengthen these conclusions?
    """
    
    critique = call_claude(swarm.agents["critique_1"], critique_prompt)
    println(critique)
    println()
    
    # -----------------------------------------------------------------------
    
    println("PHASE 7: Synthesis")
    println("=" ^ 80)
    
    synthesis_prompt = """
    Synthesize this research into a mini summary:
    
    LITERATURE: $(lit_summary)
    
    HYPOTHESIS: $(hypothesis)
    
    RESULTS: $(mock_results)
    
    ANALYSIS: $(analysis)
    
    CRITIQUE: $(critique)
    
    Write a structured research summary with: question, hypothesis, finding,
    limitation, future direction. Under 200 words.
    """
    
    final_summary = call_claude(swarm.agents["synthesis_1"], synthesis_prompt)
    println(final_summary)
    
    return final_summary
end

# Uncomment to run (will make ~7 API calls):
# final_research = run_discovery_workflow(discovery_swarm)

println("Research workflow defined (not executed to save costs)")
println("Uncomment the last line to run the full workflow")
```
