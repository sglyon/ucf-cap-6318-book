## Question 1: Mixed Strategy Nash Equilibria

**Format**: Show student a normal form game, ask them to find all mixed strategy Nash equilibria


### Variation 1.1: Battle of the Sexes (Coordination Game)

Story: Alex and Bailey are a couple planning their Friday evening. Alex prefers going to a Boxing Match, while Bailey prefers going to the Opera. However, both would rather be together at their less-preferred event than be apart. They
must decide simultaneously without communicating.

Payoffs (utility points):
- Being together at your preferred event: 2 points
- Being together at your less-preferred event: 1 point
- Being apart: 0 points for both

Normal Form Game:

|                    | Bailey: Boxing Match | Bailey: Opera |
|--------------------|----------------------|---------------|
| Alex: Boxing Match | (2, 1)               | (0, 0)        |
| Alex: Opera        | (0, 0)               | (1, 2)        |

Note: Payoffs shown as (Alex, Bailey)

Question: Find ALL Nash equilibria (both pure and mixed strategies). For the mixed strategy equilibrium, what is the probability each player chooses their preferred event? What are the expected payoffs?

#### Expected Answer

- Pure Strategy NE: (Boxing, Boxing) with payoffs (2,1) and (Opera, Opera) with payoffs (1,2)
    - At (Boxing, Boxing): Alex gets 2, wouldn't switch to 0. Bailey gets 1, wouldn't switch to 0. ✓
    - At (Opera, Opera): Alex gets 1, wouldn't switch to 0. Bailey gets 2, wouldn't switch to 0. ✓
- Mixed Strategy NE: Let p = Prob(Alex chooses Boxing), q = Prob(Bailey chooses Boxing)
    - For Bailey to be indifferent:
        - $E_B(\text{Boxing}|p) = 1p + 0(1-p) = p$
        - $E_B(\text{Opera}|p) = 0p + 2(1-p) = 2 - 2p$
        - Set equal: $p = 2 - 2p \Rightarrow 3p = 2 \Rightarrow p^* = 2/3$
    - For Alex to be indifferent:
        - $E_A(\text{Boxing}|q) = 2q + 0(1-q) = 2q$
        - $E_A(\text{Opera}|q) = 0q + 1(1-q) = 1 - q$
        - Set equal: $2q = 1 - q \Rightarrow 3q = 1 \Rightarrow q^* = 1/3$
    - Expected payoffs: $E_A = 2(2/3)(1/3) + 0 + 0 + 1(1/3)(2/3) = 4/9 + 2/9 = 6/9 = 2/3$
    - By symmetry, $E_B = 2/3$
- Answer: Three Nash equilibria exist:
    1. (Boxing, Boxing) - both get their less/more preferred in coordination
    2. (Opera, Opera) - both get their more/less preferred in coordination
    3. Mixed: Alex chooses Boxing with prob 2/3, Bailey chooses Boxing with prob 1/3, both expect 2/3 utilityk

### Variation 1.2: Hawk-Dove Game (Conflict over Resources)

Story: Two animals (or firms, or nations) encounter a valuable resource worth V = 12 units. Each can play Hawk (aggressive: fight for entire resource) or Dove (passive: share or retreat). If both play Hawk, they fight and each pays a
cost C = 18 units (injuries or damage). If one plays Hawk and other plays Dove, Hawk gets everything and Dove gets nothing. If both play Dove, they share equally.

Payoffs:
- (Hawk, Hawk): Fight, each gets V/2 - C/2 = 6 - 9 = -3
- (Hawk, Dove): Hawk gets V = 12, Dove gets 0
- (Dove, Hawk): Dove gets 0, Hawk gets V = 12
- (Dove, Dove): Share peacefully, each gets V/2 = 6

Normal Form Game:

|                | Player 2: Hawk | Player 2: Dove |
|----------------|----------------|----------------|
| Player 1: Hawk | (-3, -3)       | (12, 0)        |
| Player 1: Dove | (0, 12)        | (6, 6)         |

Note: Payoffs shown as (Player 1, Player 2)

Question: Find ALL Nash equilibria (pure and mixed). For the mixed strategy equilibrium, what is the probability each player plays Hawk? When is fighting (mutual Hawk) more likely - when the resource is very valuable or when fighting
costs are very high?

**Expected Answer**

- Pure Strategy NE:
    - (Hawk, Dove) with payoffs (12, 0): P1 gets 12 (wouldn't switch to 0 or 6), P2 gets 0 (would switch to 6 from Dove!). Wait, let me check: if P2 switches from Dove to Hawk, gets -3 instead of 0. So wouldn't switch. But would switch
    to... wait, P2 is choosing Dove. If switches to Hawk, outcome is (Hawk, Hawk) = (-3, -3). Since -3 < 0, P2 won't switch. ✓
    - (Dove, Hawk) with payoffs (0, 12): By symmetry, also NE. ✓
- Mixed Strategy: Let p = Prob(P1 plays Hawk), q = Prob(P2 plays Hawk)
    - For P2 to be indifferent:
      - $E_{P2}(\text{Hawk}|p) = -3p + 12(1-p) = 12 - 15p$
      - $E_{P2}(\text{Dove}|p) = 0p + 6(1-p) = 6 - 6p$
      - Set equal: $12 - 15p = 6 - 6p \Rightarrow 6 = 9p \Rightarrow p^* = 2/3$
    - By symmetry, $q^* = 2/3$
    - Expected payoff: $E = (-3)(2/3)(2/3) + 12(2/3)(1/3) + 0(1/3)(2/3) + 6(1/3)(1/3) = -4/3 + 8/3 + 0 + 2/3 = 6/3 = 2$
- Answer:
    - Pure NE: (Hawk, Dove) and (Dove, Hawk)
    - Mixed NE: Both play Hawk with probability 2/3, both expect payoff of 2
- General insight: As V increases (resource more valuable) or C decreases (fighting cheaper), probability of Hawk increases

### Variation 1.3: Penalty Kick in Soccer

Story: A striker is taking a penalty kick against a goalkeeper. The striker can shoot Left or Right. The goalkeeper must simultaneously dive Left or Right. Based on analysis of professional soccer data, we know the success rates: if the
goalkeeper guesses correctly, they save the penalty 60% of the time. If they guess wrong, the striker scores 90% of the time.

Payoffs (simplified to: 1 = goal scored, 0 = save):
- Shooter shoots direction, GK dives same direction: GK saves (Striker: 0, GK: 1)
- Shooter shoots direction, GK dives opposite: Striker scores (Striker: 1, GK: 0)

Normal Form Game:

|                      | Goalkeeper: Dive Left | Goalkeeper: Dive Right |
|----------------------|-----------------------|------------------------|
| Striker: Shoot Left  | (0, 1)                | (1, 0)                 |
| Striker: Shoot Right | (1, 0)                | (0, 1)                 |

Note: Payoffs shown as (Striker, Goalkeeper)

Question: Find the mixed strategy Nash equilibrium. What is the probability the striker shoots left? Should the striker shoot more often to their stronger side, or randomize equally? What is the expected probability of scoring a goal in
equilibrium?

**Expected Answer**

- This is a symmetric zero-sum game (classic matching pennies structure)
- No pure strategy NE exists (can verify each cell)
- Let p = Prob(Striker shoots Left), q = Prob(GK dives Left)
- For Goalkeeper to be indifferent:
    - $E_{GK}(\text{Dive Left}|p) = 1p + 0(1-p) = p$
    - $E_{GK}(\text{Dive Right}|p) = 0p + 1(1-p) = 1 - p$
    - Set equal: $p = 1 - p \Rightarrow p^* = 1/2$
- By symmetry of the game, $q^* = 1/2$
- Expected payoff for Striker: $0(1/4) + 1(1/4) + 1(1/4) + 0(1/4) = 1/2$
- Expected payoff for GK: $1/2$ (saves half)
- Answer:
    - Both players randomize 50-50
    - Striker expects to score 50% of the time
    - Key insight: Even if striker has a stronger side in reality, they must randomize equally in equilibrium. If they shot left 60% of the time, GK would dive left more often, reducing scoring probability. The 50-50 mix makes GK indifferent and maximizes striker's expected success.

### Variation 4.4: Weak Dominance Checker (Game Theory)

**Context**: You're analyzing a game to identify weakly dominated strategies. A strategy is weakly dominated if there exists another strategy that is at least as good against all opponent strategies, and strictly better against at least one.

````julia
using GameTheory

function is_weakly_dominated(game::NormalFormGame, player_index::Int, strategy::Int)
    # TODO: Complete this function
    # Return true if the given strategy is weakly dominated, false otherwise
    #
    # A strategy S is weakly dominated if there exists another strategy S' such that:
    # - Payoff(S') >= Payoff(S) for ALL opponent strategies
    # - Payoff(S') > Payoff(S) for AT LEAST ONE opponent strategy
    #
    # Steps:
    # 1. Get player's payoff matrix
    # 2. Get payoffs for the strategy being tested
    # 3. For each OTHER strategy, check if it weakly dominates
    # 4. Return true if any strategy weakly dominates, false otherwise
    
end
````

**Expected Answer**:

```julia
function is_weakly_dominated(game::NormalFormGame, player_index::Int, strategy::Int)
    payoffs = game.players[player_index].payoff_array
    
    # Get payoffs for the strategy being tested
    if player_index == 1
        strategy_payoffs = payoffs[strategy, :]
    else
        strategy_payoffs = payoffs[:, strategy]
    end
    
    num_strategies = player_index == 1 ? size(payoffs, 1) : size(payoffs, 2)
    
    # Check each other strategy
    for other_strat in 1:num_strategies
        if other_strat == strategy
            continue
        end
        
        # Get payoffs for the other strategy
        if player_index == 1
            other_payoffs = payoffs[other_strat, :]
        else
            other_payoffs = payoffs[:, other_strat]
        end
        
        # Check weak dominance: at least as good everywhere, strictly better somewhere
        at_least_as_good = all(other_payoffs .>= strategy_payoffs)
        strictly_better_somewhere = any(other_payoffs .> strategy_payoffs)
        
        if at_least_as_good && strictly_better_somewhere
            return true  # Found a weakly dominating strategy
        end
    end
    
    return false  # No strategy weakly dominates this one
end
```

**Key points to check**:
- Correctly extracts payoffs for the strategy being tested
- Iterates through all OTHER strategies (skips the strategy itself)
- Correctly implements weak dominance test:
  - `>=` for all opponent strategies (at least as good)
  - `>` for at least one opponent strategy (strictly better somewhere)
- Returns boolean
- Handles player 1 vs player 2 indexing correctly
- **Acceptable variations**: Could use vector operations differently, could return which strategy dominates it, could check for strict dominance instead
